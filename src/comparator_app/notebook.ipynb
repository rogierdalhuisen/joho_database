```python
import os
import requests
from dotenv import load_dotenv
import json

# Laad je .env bestand
load_dotenv()

# Haal de endpoints en token op
RELATIES_URL = os.getenv('ASSUPORTAL_RELATIES')
CONTRACTEN_URL = os.getenv('ASSUPORTAL_CONTRACTEN')
BEARER_TOKEN = os.getenv('ASSUPORTAL_API_TOKEN')  # Of hoe je het token ook genoemd hebt

print(f"Relaties URL: {RELATIES_URL}")
print(f"Contracten URL: {CONTRACTEN_URL}")
print(f"Token aanwezig: {'Ja' if BEARER_TOKEN else 'Nee'}")
```

```python
# Configureer de headers met je Bearer token
headers = {
    'Authorization': f'Bearer {BEARER_TOKEN}',
    'Content-Type': 'application/json'
}

# Configureer pagination parameters (start met een kleine page size voor testen)
params = {
    'page': 1,
    'size': 10  # Klein aantal voor eerste test
}
```

```python
test = RELATIES_URL + '/' + str(10690)
```

```python
# Maak de API call naar het Relaties endpoint
try:
    print("Making API call to Relaties endpoint...")
    response_relaties = requests.get(RELATIES_URL, headers=headers, params=params, timeout=30)

    # Check of de request succesvol was
    response_relaties.raise_for_status()

    print(f"✓ Success! Status code: {response_relaties.status_code}")

    # Parse de JSON response
    data_relaties = response_relaties.json()

    # Print de volledige response (mooi geformatteerd)
    print("\n--- Complete JSON Response ---")
    print(json.dumps(data_relaties, indent=2, ensure_ascii=False))
except requests.exceptions.RequestException as e:
    print(f"✗ Error: {e}")
    if hasattr(e.response, 'text'):
        print(f"Response text: {e.response.text}")
```

```python
detail_test_ids = [1, 2, 3, 4, 5]  # De IDs uit je response hierboven

print("=== TESTING DETAIL ENDPOINTS ===\n")

for rel_id in detail_test_ids:
    try:
        url = f"{RELATIES_URL}/{rel_id}"
        response = requests.get(url, headers=headers, timeout=30)
        response.raise_for_status()

        data = response.json()['data']

        print(f"ID {rel_id}:")
        print(f"  - Has personen: {len(data.get('personen', []))} items")
        print(f"  - Has adressen: {len(data.get('adressen', []))} items")
        print(f"  - Has email_adressen: {len(data.get('email_adressen', []))} items")
        print(f"  - Has contracten: {len(data.get('contracten', []))} items")
        print(f"  - Top-level keys: {len(data.keys())} fields\n")

    except Exception as e:
        print(f"ID {rel_id}: {e}\n")

# Test ook verschillende Contracten
print("\n=== TESTING CONTRACTEN VARIATIONS ===\n")

contracten_params = {'page': 500, 'size': 5}  # Andere pagina
response = requests.get(CONTRACTEN_URL, headers=headers, params=contracten_params, timeout=30)

contracten_data = response.json()
for contract in contracten_data['data'][:3]:  # Eerste 3
    print(f"Contract {contract['id']}:")
    print(f"  - Status: '{contract.get('status')}'")
    print(f"  - Type: {contract.get('type')}")
    print(f"  - Has relatie_id: {contract.get('relatie_id') is not None}")
    print(f"  - Keys: {sorted(contract.keys())}\n")
```
